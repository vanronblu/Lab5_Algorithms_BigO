```python
# Lab 5 Appendix - Exercise 1: Emergency Dispatch Record Search

LAST_NAME = "GERMAN"
STUDENT_ID = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

# Derived student-specific values 
id_digits = [int(ch) for ch in STUDENT_ID if ch.isdigit()]            # [3,6,4,3,3,6]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]       # U,N,I,Q,...

SEED_DIGIT = id_digits[-1]                                            # last digit of STUDENT_ID
ID_SUM = sum(id_digits)                                               # sum of STUDENT_ID digits
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(artist_letters)
ARTIST_SUM = sum(ord(ch) - 64 for ch in artist_letters)               # A=1 ... Z=26

print("=== STUDENT-SPECIFIC INPUTS ===")
print("LAST_NAME       :", LAST_NAME)
print("STUDENT_ID      :", STUDENT_ID)
print("FAVORITE_ARTIST :", FAVORITE_ARTIST)
print("SEED_DIGIT      :", SEED_DIGIT)
print("ID_SUM          :", ID_SUM)
print("NAME_LENGTH     :", NAME_LENGTH)
print("ARTIST_LENGTH   :", ARTIST_LENGTH)
print("ARTIST_SUM      :", ARTIST_SUM)

# Original Dataset (sorted, unique incident numbers) 
NUM_RECORDS = 15
original = []
value = 1000
for i in range(NUM_RECORDS):
    letter_value = ord(artist_letters[i % ARTIST_LENGTH]) - 64
    value = value + ID_SUM + (letter_value % SEED_DIGIT) + 1          
    original.append(value)

print("\n=== 1. ORIGINAL DATASET (sorted) ===")
print(original)

# Rotated Dataset 
rotation_point = SEED_DIGIT
rotated = original[rotation_point:] + original[:rotation_point]

print("\n=== 2. ROTATED DATASET (rotation point =", rotation_point, ") ===")
print(rotated)

# Target Incident 
target_index = (ID_SUM + ARTIST_SUM) % NUM_RECORDS
target = original[target_index]

print("\n=== 3. TARGET INCIDENT ===")
print("Target index rule : (ID_SUM + ARTIST_SUM) % NUM_RECORDS =",
      "(" + str(ID_SUM), "+", str(ARTIST_SUM) + ") %", NUM_RECORDS, "=", target_index)
print("Target incident   :", target)

# Search algorithms
def rotated_binary_search(arr, target):
    # Modified binary search for a rotated sorted list.
    # Every comparison between values is counted.
    comparisons = 0
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1                       # arr[mid] == target
        if arr[mid] == target:
            return mid, comparisons
        comparisons += 1                       # which half is sorted?
        if arr[low] <= arr[mid]:               # left half is sorted
            comparisons += 1                   # is target inside left half?
            if arr[low] <= target < arr[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                                  # right half is sorted
            comparisons += 1                   # is target inside right half?
            if arr[mid] < target <= arr[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons

def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

# Search Results and Comparison Counts
pos_bs, comp_bs = rotated_binary_search(rotated, target)
pos_ls, comp_ls = linear_search(rotated, target)

print("\n=== 4. SEARCH RESULTS AND COMPARISON COUNTS ===")
if pos_bs != -1:
    print("Rotated Binary Search : FOUND at index", pos_bs, "| comparisons =", comp_bs)
else:
    print("Rotated Binary Search : NOT FOUND | comparisons =", comp_bs)
if pos_ls != -1:
    print("Linear Search         : FOUND at index", pos_ls, "| comparisons =", comp_ls)
else:
    print("Linear Search         : NOT FOUND | comparisons =", comp_ls)

# Extra test: a target that is not in the list
missing = target + 1
pos_m1, comp_m1 = rotated_binary_search(rotated, missing)
pos_m2, comp_m2 = linear_search(rotated, missing)
print("\nNot-found test (target", missing, "):")
print("Rotated Binary Search :", "NOT FOUND" if pos_m1 == -1 else pos_m1, "| comparisons =", comp_m1)
print("Linear Search         :", "NOT FOUND" if pos_m2 == -1 else pos_m2, "| comparisons =", comp_m2)

# Comparison with Linear Search 
print("\n=== 5. COMPARISON WITH LINEAR SEARCH ===")
print("Both searches found the same position:", pos_bs == pos_ls)
print("Binary comparisons :", comp_bs)
print("Linear comparisons :", comp_ls)
if comp_bs < comp_ls:
    print("Rotated binary search used", comp_ls - comp_bs, "fewer comparisons on this target.")
else:
    print("Linear search used", comp_bs - comp_ls, "fewer comparisons on this target.")
print("Note: binary search counts 3 checks per loop, so its advantage grows with larger lists.")

# Time and Space Complexity
print("\n=== 6. TIME AND SPACE COMPLEXITY ===")
print(f"{'Algorithm':<24} {'Best':<7} {'Average':<10} {'Worst':<10} {'Space'}")
print(f"{'Rotated Binary Search':<24} {'O(1)':<7} {'O(log n)':<10} {'O(log n)':<10} O(1)")
print(f"{'Linear Search':<24} {'O(1)':<7} {'O(n)':<10} {'O(n)':<10} O(1)")

# FINAL SEARCH REPORT 
print("\n" + "=" * 62)
print("FINAL EMERGENCY DISPATCH SEARCH REPORT")
print("=" * 62)
print("Student            :", LAST_NAME, "|", STUDENT_ID, "|", FAVORITE_ARTIST)
print("Records            :", NUM_RECORDS)
print("Rotation point     :", rotation_point)
print("Target incident    :", target)
print("Position (rotated) :", pos_bs)
print("Binary comparisons :", comp_bs)
print("Linear comparisons :", comp_ls)
print("Binary Search      : O(log n) time, O(1) space")
print("Linear Search      : O(n) time, O(1) space")
print("=" * 62)
```

```python
# Lab 5 Appendix - Exercise 2: University Enrollment Records

LAST_NAME = "GERMAN"
STUDENT_ID = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

# Derived student-specific values
id_digits = [int(ch) for ch in STUDENT_ID if ch.isdigit()]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(artist_letters)

print("=== STUDENT-SPECIFIC INPUTS ===")
print("SEED_DIGIT    :", SEED_DIGIT)
print("ID_SUM        :", ID_SUM)
print("NAME_LENGTH   :", NAME_LENGTH)
print("ARTIST_LENGTH :", ARTIST_LENGTH)

# Department A dataset 
dept_a = []
for i in range(8):
    dept_a.append(ID_SUM + (SEED_DIGIT + 1) * i)

print("\n=== 1. DEPARTMENT A DATASET ===")
print(dept_a)

# Department B dataset
dept_b = []
for i in range(10):
    dept_b.append(ARTIST_LENGTH + (NAME_LENGTH + 1) * i)

print("\n=== 2. DEPARTMENT B DATASET ===")
print(dept_b)

# Merge algorithm (two-pointer)
def merge_sorted(a, b):
    i = 0
    j = 0
    merged = []
    comparisons = 0
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    while i < len(a):         
        merged.append(a[i])
        i += 1
    while j < len(b):        
        merged.append(b[j])
        j += 1
    return merged, comparisons

# Consolidated dataset
consolidated, comparisons = merge_sorted(dept_a, dept_b)

print("\n=== 3. CONSOLIDATED DATASET ===")
print(consolidated)

# Number of records before and after 
print("\n=== 4. NUMBER OF RECORDS BEFORE AND AFTER CONSOLIDATION ===")
print("Department A records :", len(dept_a))
print("Department B records :", len(dept_b))
print("Total before merging :", len(dept_a) + len(dept_b))
print("Total after merging  :", len(consolidated))

# Comparison count
print("\n=== 5. COMPARISON COUNT ===")
print("Comparisons performed        :", comparisons)
print("Best case  (min of n, m)     :", min(len(dept_a), len(dept_b)))
print("Worst case (n + m - 1)       :", len(dept_a) + len(dept_b) - 1)

# Ordering verification 
is_ordered = True
for k in range(len(consolidated) - 1):
    if consolidated[k] > consolidated[k + 1]:
        is_ordered = False
same_count = len(consolidated) == len(dept_a) + len(dept_b)
same_values = consolidated == sorted(dept_a + dept_b)

print("\n=== 6. ORDERING VERIFICATION ===")
print("Ascending order        :", "PASSED" if is_ordered else "FAILED")
print("Record count correct   :", "PASSED" if same_count else "FAILED")
print("Matches sorted(A + B)  :", "PASSED" if same_values else "FAILED")

# Time and space complexity 
print("\n=== 7. TIME AND SPACE COMPLEXITY ===")
print("n =", len(dept_a), "(Dept A), m =", len(dept_b), "(Dept B)")
print("Time complexity  : O(n + m) - each record is placed once")
print("Space complexity : O(n + m) - a new list holds all merged records")

# FINAL CONSOLIDATED ENROLLMENT LIST 
print("\n" + "=" * 62)
print("FINAL CONSOLIDATED ENROLLMENT LIST")
print("=" * 62)
print("Student  :", LAST_NAME, "|", STUDENT_ID, "|", FAVORITE_ARTIST)
print("Dept A   :", dept_a)
print("Dept B   :", dept_b)
print("Merged   :", consolidated)
print("Records  :", len(dept_a), "+", len(dept_b), "=", len(consolidated))
print("Comparisons :", comparisons)
print("Verified ordered :", is_ordered and same_count and same_values)
print("Complexity  : O(n + m) time, O(n + m) space")
print("=" * 62)
```

```python
# Lab 5 Appendix - Exercise 3: E-Commerce Order Processing
import random
import time

# Student-specific inputs
LAST_NAME = "GERMAN"
STUDENT_ID = "TUPM-36-4336"
FAVORITE_ARTIST = "UNIQUE SALONGA"

# Derived student-specific values
id_digits = [int(ch) for ch in STUDENT_ID if ch.isdigit()]
artist_letters = [ch for ch in FAVORITE_ARTIST if ch.isalpha()]

SEED_DIGIT = id_digits[-1]
ID_SUM = sum(id_digits)
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(artist_letters)
ARTIST_SUM = sum(ord(ch) - 64 for ch in artist_letters)

print("=== STUDENT-SPECIFIC INPUTS ===")
print("SEED_DIGIT:", SEED_DIGIT, "| ID_SUM:", ID_SUM, "| NAME_LENGTH:", NAME_LENGTH,
      "| ARTIST_LENGTH:", ARTIST_LENGTH, "| ARTIST_SUM:", ARTIST_SUM)

# Generated Order Dataset 
NUM_ORDERS = 20
orders = []
for i in range(NUM_ORDERS):
    letter_value = ord(artist_letters[i % ARTIST_LENGTH]) - 64
    order_no = 1000 + (ID_SUM * (i + 1) + SEED_DIGIT * letter_value) % 500
    while order_no in orders:          # keep every order number unique
        order_no += 1
    orders.append(order_no)

print("\n=== 1. GENERATED ORDER DATASET ===")
print(orders)

# Sorted, Reverse, and Random Inputs (same orders)
sorted_input = sorted(orders)
reverse_input = sorted(orders, reverse=True)
random_input = orders[:]
random.seed(int("".join(str(d) for d in id_digits)))     # seed = 364336 (reproducible)
random.shuffle(random_input)

conditions = [("SORTED", sorted_input), ("REVERSE", reverse_input), ("RANDOM", random_input)]

print("\n=== 2. SORTED, REVERSE, AND RANDOM INPUTS ===")
for name, data in conditions:
    print(f"{name:<8}: {data}")

# Sorting algorithms
# Movements: swaps for Bubble/Selection, shifts for Insertion.
def bubble_sort(data):
    arr = data[:]
    comparisons = 0
    movements = 0
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            comparisons += 1
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 1
                swapped = True
        if not swapped:                
            break
    return arr, comparisons, movements

def selection_sort(data):
    arr = data[:]
    comparisons = 0
    movements = 0
    n = len(arr)
    for i in range(n - 1):
        min_index = i
        for j in range(i + 1, n):
            comparisons += 1
            if arr[j] < arr[min_index]:
                min_index = j
        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 1
    return arr, comparisons, movements

def insertion_sort(data):
    arr = data[:]
    comparisons = 0
    movements = 0
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break
        arr[j + 1] = key
    return arr, comparisons, movements

algorithms = [("Bubble Sort", bubble_sort),
              ("Selection Sort", selection_sort),
              ("Insertion Sort", insertion_sort)]

# Run everything
RUNS = 1000
results = {}   
correct_order = sorted(orders)

for algo_name, algo in algorithms:
    for cond_name, data in conditions:
        result, comps, moves = algo(data)
        start = time.perf_counter()
        for _ in range(RUNS):
            algo(data)
        end = time.perf_counter()
        avg_us = (end - start) / RUNS * 1_000_000
        results[(algo_name, cond_name)] = {
            "comparisons": comps, "movements": moves,
            "time_us": avg_us, "verified": result == correct_order}

# Comparison Counts 
print("\n=== 3. COMPARISON COUNTS ===")
print(f"{'Algorithm':<16} {'SORTED':<8} {'REVERSE':<9} {'RANDOM'}")
for algo_name, algo in algorithms:
    print(f"{algo_name:<16} {results[(algo_name, 'SORTED')]['comparisons']:<8} "
          f"{results[(algo_name, 'REVERSE')]['comparisons']:<9} "
          f"{results[(algo_name, 'RANDOM')]['comparisons']}")

# Movement Counts 
print("\n=== 4. MOVEMENT COUNTS (swaps / shifts) ===")
print(f"{'Algorithm':<16} {'SORTED':<8} {'REVERSE':<9} {'RANDOM'}")
for algo_name, algo in algorithms:
    print(f"{algo_name:<16} {results[(algo_name, 'SORTED')]['movements']:<8} "
          f"{results[(algo_name, 'REVERSE')]['movements']:<9} "
          f"{results[(algo_name, 'RANDOM')]['movements']}")

# Execution Times 
print("\n=== 5. EXECUTION TIMES (microseconds, average of", RUNS, "runs) ===")
print(f"{'Algorithm':<16} {'SORTED':<10} {'REVERSE':<10} {'RANDOM'}")
for algo_name, algo in algorithms:
    print(f"{algo_name:<16} {results[(algo_name, 'SORTED')]['time_us']:<10.2f} "
          f"{results[(algo_name, 'REVERSE')]['time_us']:<10.2f} "
          f"{results[(algo_name, 'RANDOM')]['time_us']:.2f}")

#  Sorting Verification 
print("\n=== 6. SORTING VERIFICATION ===")
all_ok = True
for algo_name, algo in algorithms:
    for cond_name, data in conditions:
        ok = results[(algo_name, cond_name)]["verified"]
        if not ok:
            all_ok = False
        print(f"{algo_name:<16} {cond_name:<8} {'PASSED' if ok else 'FAILED'}")
print("All results correctly sorted:", all_ok)

#  Big O Analysis
print("\n=== 7. BIG O ANALYSIS ===")
print(f"{'Algorithm':<16} {'Best':<8} {'Average':<9} {'Worst':<9} {'Space'}")
print(f"{'Bubble Sort':<16} {'O(n)':<8} {'O(n^2)':<9} {'O(n^2)':<9} O(1)   (with early-exit check)")
print(f"{'Selection Sort':<16} {'O(n^2)':<8} {'O(n^2)':<9} {'O(n^2)':<9} O(1)")
print(f"{'Insertion Sort':<16} {'O(n)':<8} {'O(n^2)':<9} {'O(n^2)':<9} O(1)")
n = NUM_ORDERS
print("\nFor n =", n, ": n - 1 =", n - 1, "and n(n - 1)/2 =", n * (n - 1) // 2)

#  Best-, Average-, and Worst-Case Observations 
print("\n=== 8. BEST-, AVERAGE-, AND WORST-CASE OBSERVATIONS ===")
for algo_name, algo in algorithms:
    b = results[(algo_name, "SORTED")]
    a = results[(algo_name, "RANDOM")]
    w = results[(algo_name, "REVERSE")]
    print(algo_name)
    print("  Best  (SORTED) :", b["comparisons"], "comparisons,", b["movements"], "movements")
    print("  Avg   (RANDOM) :", a["comparisons"], "comparisons,", a["movements"], "movements")
    print("  Worst (REVERSE):", w["comparisons"], "comparisons,", w["movements"], "movements")

print("\nLowest total operations (comparisons + movements) per condition:")
for cond_name, data in conditions:
    best_algo = None
    best_total = None
    for algo_name, algo in algorithms:
        r = results[(algo_name, cond_name)]
        total = r["comparisons"] + r["movements"]
        if best_total is None or total < best_total:
            best_total = total
            best_algo = algo_name
    print(f"  {cond_name:<8}: {best_algo} ({best_total} operations)")

# FINAL SORTING PERFORMANCE REPORT 
print("\n" + "=" * 70)
print("FINAL SORTING PERFORMANCE REPORT")
print("=" * 70)
print("Student :", LAST_NAME, "|", STUDENT_ID, "|", FAVORITE_ARTIST)
print("Orders  :", NUM_ORDERS, "| Random seed:", int("".join(str(d) for d in id_digits)))
print("-" * 70)
print(f"{'ALGORITHM':<16} {'INPUT':<8} {'COMP':<6} {'MOVES':<6} {'TIME(us)':<10} VERIFIED")
for algo_name, algo in algorithms:
    for cond_name, data in conditions:
        r = results[(algo_name, cond_name)]
        print(f"{algo_name:<16} {cond_name:<8} {r['comparisons']:<6} {r['movements']:<6} "
              f"{r['time_us']:<10.2f} {'YES' if r['verified'] else 'NO'}")
    print("-" * 70)
print("Selection Sort makes the same number of comparisons for every input.")
print("Bubble and Insertion Sort are fastest on SORTED input and slowest on REVERSE input.")
print("=" * 70)
```
